## Inventory filename characters in source corpus

```python
from pathlib import Path

source_dir = Path("corpus/00_source")

filename_characters = set()

for path in source_dir.iterdir():
    if path.is_file():
        filename_characters.update(path.name)

filename_characters
```

## Normalise source corpus filenames

- lower the case
- replace ' ' and ',' with '_'
- replace '.' with '_' except when it precedes the file extension
- replace 'ã' with 'a'
- replace 'ç' with 'c'
- replace 'ê' with 'e'

```python
from pathlib import Path

source_dir = Path("corpus/00_source")

rename_log = []

for path in source_dir.iterdir():
    if not path.is_file():
        continue

    stem = path.stem
    suffix = path.suffix.lower()

    new_stem = (
        stem.lower()
        .replace(" ", "_")
        .replace(",", "_")
        .replace(".", "_")
        .replace("ã", "a")
        .replace("ç", "c")
        .replace("ê", "e")
    )

    new_path = path.with_name(f"{new_stem}{suffix}")

    if path.name != new_path.name:
        if new_path.exists():
            raise FileExistsError(f"Cannot rename {path.name}: {new_path.name} already exists")

        path.rename(new_path)
        rename_log.append((path.name, new_path.name))

rename_log
```

## Count source files by Brazilian state code

```python
from collections import Counter
from pathlib import Path
import pandas as pd

source_dir = Path("corpus/00_source")

state_counts = Counter(
    path.name.split("_", 1)[0]
    for path in source_dir.iterdir()
    if path.is_file()
)

state_counts_df = (
    pd.Series(state_counts, name="file_count")
    .rename_axis("state_code")
    .sort_index()
    .reset_index()
)

state_counts_df
```

## Inspect educational stage and discipline labels from filenames

```python
from pathlib import Path

source_dir = Path("corpus/00_source")

stage_discipline_labels = {
    filename.stem[filename.stem.find("_") + 1:filename.stem.rfind("_")]
    for filename in source_dir.glob("*.txt")
    if filename.stem.count("_") >= 2
}

stage_discipline_labels = set(sorted(stage_discipline_labels))

stage_discipline_labels
```

## Normalise double underscores in source filenames

```python
from pathlib import Path

source_dir = Path("corpus/00_source")

renamed_files = []

for path in source_dir.glob("*.txt"):
    new_name = path.name
    while "__" in new_name:
        new_name = new_name.replace("__", "_")

    if new_name != path.name:
        new_path = path.with_name(new_name)

        if new_path.exists():
            raise FileExistsError(f"Cannot rename {path.name} to {new_name}: target already exists.")

        path.rename(new_path)
        renamed_files.append((path.name, new_name))

renamed_files
```

## Organise source files by Brazilian state code

```python
from collections import Counter
from pathlib import Path
import shutil

source_dir = Path("corpus/00_source")
per_state_dir = Path("corpus/00_source_per_state")

per_state_dir.mkdir(parents=True, exist_ok=True)

copied_files = []

for path in sorted(source_dir.iterdir()):
    if not path.is_file():
        continue

    state_code = path.name.split("_", 1)[0].lower()
    state_dir = per_state_dir / state_code
    state_dir.mkdir(parents=True, exist_ok=True)

    target_path = state_dir / path.name
    shutil.copy2(path, target_path)
    copied_files.append((path.name, state_code, str(target_path)))

per_state_counts_df = (
    pd.Series(Counter(state for _, state, _ in copied_files), name="file_count")
    .rename_axis("state_code")
    .sort_index()
    .reset_index()
)

per_state_counts_df
```

## Create source-file inventory by Brazilian state

```python
from pathlib import Path
import pandas as pd

source_per_state_dir = Path("corpus/00_source_per_state")

records = []

for state_dir in sorted(source_per_state_dir.iterdir()):
    if not state_dir.is_dir():
        continue

    state = state_dir.name

    for path in sorted(state_dir.iterdir()):
        if path.is_file():
            records.append(
                {
                    "filename": path.name,
                    "state": state,
                    "filepath": path.as_posix(),
                }
            )

df_brazilian_educational_guidelines = pd.DataFrame(
    records,
    columns=["filename", "state", "filepath"],
)

df_brazilian_educational_guidelines
```

## Add original text word counts

```python
from pathlib import Path
import re

def count_words_in_file(filepath):
    text = Path(filepath).read_text(encoding="utf-8")
    return len(re.findall(r"\b\w+\b", text, flags=re.UNICODE))

df_brazilian_educational_guidelines["word_count_original"] = (
    df_brazilian_educational_guidelines["filepath"].apply(count_words_in_file)
)

df_brazilian_educational_guidelines
```

## Collect POS tags for lemma `que`

```python
from pathlib import Path

base = Path("corpus/07_tagged")
que_tags = set()

for tagged_file in base.rglob("*.txt"):
    with tagged_file.open("r", encoding="utf-8") as f:
        for line in f:
            parts = line.strip().split("\t")
            if len(parts) >= 3 and parts[2].strip().lower() == "que":
                que_tags.add(parts[1])

que_tags
```
